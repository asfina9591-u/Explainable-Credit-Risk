```python
X = df.drop("SeriousDlqin2yrs", axis=1)
y = df["SeriousDlqin2yrs"]

print("Features:", X.shape)
print("Target:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True).round(4))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True).round(4))
```

```python
from xgboost import XGBClassifier

negative = (y_train == 0).sum()
positive = (y_train == 1).sum()

scale_pos_weight = negative / positive

print("Negative cases:", negative)
print("Positive cases:", positive)
print("Scale pos weight:", round(scale_pos_weight, 2))
```

```python
model = XGBClassifier(
    n_estimators=300,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)

print("XGBoost model trained successfully!")
```

```python
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]

print("Predictions generated successfully!")
print("Number of predictions:", len(y_pred))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_prob)

print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))
print("ROC-AUC  :", round(roc_auc, 4))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Low Risk", "High Risk"],
    yticklabels=["Low Risk", "High Risk"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Credit Risk Confusion Matrix")

plt.show()
```

```python
from sklearn.metrics import roc_curve, auc

fpr, tpr, thresholds = roc_curve(y_test, y_prob)

roc_auc_value = auc(fpr, tpr)

plt.figure(figsize=(7, 5))

plt.plot(
    fpr,
    tpr,
    label=f"XGBoost (AUC = {roc_auc_value:.3f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()

plt.show()
```

```python
customer_index = 0

customer = X_test.iloc[[customer_index]]

customer_prediction = model.predict(customer)[0]
customer_probability = model.predict_proba(customer)[0][1]

print("Customer index:", customer_index)
print("Predicted class:", customer_prediction)
print("Risk probability:", round(customer_probability * 100, 2), "%")
```

```python
customer_shap = pd.DataFrame({
    "Feature": X_test.columns,
    "Value": customer.iloc[0].values,
    "SHAP Impact": shap_values[customer_index]
})

customer_shap["Absolute Impact"] = customer_shap["SHAP Impact"].abs()

customer_shap = customer_shap.sort_values(
    "Absolute Impact",
    ascending=False
)

customer_shap.head(10)
```

```python
top_factors = customer_shap[
    customer_shap["SHAP Impact"] > 0
].head(3)

print("Top factors increasing predicted risk:")
print(top_factors[["Feature", "Value", "SHAP Impact"]])
```

```python
protective_factors = customer_shap[
    customer_shap["SHAP Impact"] < 0
].sort_values(
    "SHAP Impact",
    ascending=False
).head(3)

print("\nFactors reducing predicted risk:")
print(protective_factors[["Feature", "Value", "SHAP Impact"]])
```
